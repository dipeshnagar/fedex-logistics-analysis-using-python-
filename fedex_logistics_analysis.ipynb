## EDA- Expolatory Data Analysis

```python
df
```

```python
df['pq_first_sent_to_client_date'].isnull().value_counts()
```

```python
df['lead_time_days']=(df['delivered_to_client_date']-df['po_sent_to_vendor_date']).dt.days
df['full_process_days']=(df['delivered_to_client_date']-df['pq_first_sent_to_client_date']).dt.days
```

```python
df['freight_cost_per_KG']=df['freight_cost_(usd)']/df['weight_(kilograms)']
```

```python
df['freight_cost_per_KG'].replace(np.inf,np.nan,inplace=True)
```

```python
df.columns
```

```python
avg_shipment_cost=df.groupby(['shipment_mode'])['freight_cost_per_KG'].mean()
```

```python
ax=avg_shipment_cost.plot(kind='bar')
for i in ax.containers:
    ax.bar_label(i)

plt.title('Cost per KG by Shipment Mode' )
plt.xlabel('Shipment Mode')
plt.ylabel('Cost per KG')
plt.grid(color='black',alpha=1,linestyle='--',linewidth=0.4,axis='y')
plt.xticks(rotation=45)

plt.tight_layout()
```

Air Charter has the highest cost per KG significantly higher than other modes, indicating premium pricing for faster delivery.
Truck and Ocean are more cost-efficient options.

```python
shipment_days=df.groupby(['shipment_mode'])['lead_time_days'].mean()
```

```python
ax1=shipment_days.plot(kind='bar') 
for i in ax1.containers:
    ax1.bar_label(i)

plt.title("Average Lead Time by Shipment Mode", fontsize=15)
plt.xlabel("Shipment Mode", fontsize=13)
plt.ylabel("Average Lead Time (Days)", fontsize=12)
plt.xticks(rotation=45)
plt.grid(alpha=0.8,axis='y',linestyle='--',linewidth=0.5,color='red')
plt.tight_layout()
```

Ocean shipments have the highest lead time, indicating slower delivery but better cost efficiency. 
In contrast, truck shipments offer the fastest and most cost-effective delivery option. 
Air charter, while significantly more expensive, provides faster delivery than ocean but is still slower than truck.

```python
df['Delivery_delay'] = (df['delivered_to_client_date'] - df['scheduled_delivery_date']).dt.days
```

```python
df['month'] = df['delivered_to_client_date'].dt.month

plt.figure(figsize=(15,8))
sn.lineplot(data=df,x='month',y='Delivery_delay',hue='shipment_mode',estimator='mean',marker='o')

plt.title("Delivery Delay Trend by Shipment Mode")
plt.xlabel("Month")
plt.ylabel("Average Delay (Days)")
plt.legend(title="Shipment Mode")
plt.grid()
```

Air is the most reliable and consistent mode, delivering mostly on time with minimal delays, while Ocean shows consistent delays and high variability, making it unsuitable for urgent shipments.
Air Charter is the fastest option but highly expensive and unpredictable, often delivering significantly early (20–30 days before schedule).
Truck provides a cost-effective and faster alternative with moderate consistency, making it suitable for regional logistics.
When delivery time is critical, Air Charter is the best choice despite its cost due to its ability to deliver well ahead of schedule.

```python
vendor_cost = df.groupby('vendor')['freight_cost_per_KG'].mean().nsmallest(10)

vendor_cost.plot(kind='barh', figsize=(8,5))
plt.title("Top 10 Cheapest Vendors Average (Cost per KG)")
plt.show()
```

A few vendors offer significantly lower cost per KG creating strong cost-saving opportunities, while the rest cluster closely in price, making non-cost factors like reliability and service equally important.

```python


# Select top 10 countries (by shipment count)
top_countries = df['country'].value_counts().head(10).index

# Filter data
df_top = df[df['country'].isin(top_countries)]

# Pivot table
heatmap_data = df_top.pivot_table(
    values='Delivery_delay',
    index='country',
    columns='shipment_mode',
    aggfunc='mean'
)

# Plot
plt.figure(figsize=(10,6))
sn.heatmap(
    heatmap_data,
    annot=True,
    cmap='coolwarm',
    center=0,
    vmin=-20,
    vmax=20
)

plt.title("Delivery Delay (Days) by Country and Shipment Mode(Top 10 Countries)")
plt.xlabel("Shipment Mode")
plt.ylabel("Country")

plt.tight_layout()
plt.show()
```

Air shipments are consistently delivered before the scheduled date. For example, in Côte d’Ivoire, deliveries arrived up to 36 days earlier than planned. In most other countries, air deliveries are generally on time.

In contrast, ocean shipments tend to experience delays across most regions. However, Zimbabwe stands out, where ocean deliveries are relatively faster compared to other countries.

In South Africa, air deliveries perform exceptionally well, arriving up to 50 days early, while ocean shipments face delays of around 9–10 days.

Meanwhile, in countries like Haiti and Mozambique, deliveries across all transport modes are quite stable, typically arriving on time or within a small variation of 1–2 days early or late.
